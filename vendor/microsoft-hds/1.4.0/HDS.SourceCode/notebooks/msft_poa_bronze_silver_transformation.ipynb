#### IDMSilverIngestionService
The following arguments are used by the `IDMSilverIngestionService`:
- `workspace_name` : Name of the Fabric Workspace
- `solution_name` : Name of the DMH OneLake workload solution
- `admin_lakehouse_name`: The lakehouse name of where the administration configurations are located
- `one_lake_endpoint` : The one lake endpoint. Default is `onelake.dfs.fabric.microsoft.com`
