#### Overview
In this notebook, we will leverage the `AIEnrichmentsSilverIngestionService` in Healthcare data solutions library to move enrichment data  (.json/.ndjson files) from the bronze lakehouse table (AIEnrichments) into 5 Silver lakehouse tables (TextEnrichments, ObjectEnrichment, EmbeddingEnrichments, Segmentation_2dEnrichments and Segmentation_3dEnrichments).

##### Notebook Job Scheduling:
We recommend scheduling this notebook job to run every 4 hours. Initial run may not have data to consume due to concurrent and dependent jobs, leading to latency. Adjusting the frequency of higher layer jobs can reduce this latency.
