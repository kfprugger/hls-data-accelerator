# Overview
In this notebook, we will leverage the `SDOHSilverIngestionService` in Healthcare data solutions library to transform and ingest the `SDOH data` from the bronze lakehouse into the silver lakehouse. By default, you are not expected to make any changes to this notebook.

## Usage 
This notebook should be linked to bronze lakehouse.

## SDOHSilverIngestionService
The `SDOHSilverIngestionService` takes in the following arguments:
 - `spark`: spark session,
 - `workspace_name`: Name of the Fabric Workspace,
 - `solution_name`: Name of the HDS-Healthcare data solutions OneLake workload solution,
 - `admin_lakehouse_name`: The lakehouse name where the administration configurations are located,
 - `one_lake_endpoint` (str): The one lake endpoint. Default is `onelake.dfs.fabric.microsoft.com`.
