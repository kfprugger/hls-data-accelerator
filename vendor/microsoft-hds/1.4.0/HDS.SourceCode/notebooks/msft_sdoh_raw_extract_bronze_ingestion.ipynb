# Overview
In this notebook, we will leverage the `SDOHBronzeIngestionService` in Healthcare data solutions library to ingest the `public SDOH datasets` which are deployed as part of sample data. We will ingest the data into `msft_bronze` lakehouse, creating or updating tables for `SdohLayout` and `SdohDataSetMetadata`, as well as generating new tables for all the data sheets.

## SDOHBronzeIngestionService
The `SDOHBronzeIngestionService` takes in the following arguments:
 - `spark`: spark session
 - `workspace_name`: Name of the Fabric Workspace,
 - `solution_name`: Name of the HDS-Healthcare data solutions OneLake workload solution,
 - `admin_lakehouse_name:`: The lakehouse name where the administration configurations are located,
 - `one_lake_endpoint` (str): The one lake endpoint. Default is `onelake.dfs.fabric.microsoft.com`
