# Overview
In this notebook, we will leverage the `MetadataToFhirConvertor` in Healthcare data solutions library to convert the Dicom metadata present in the `ImagingDicom` table to `ImagingStudy` Fhir4 resource and save the output in the form of ndjson files.

## MetadataToFhirConvertor
The `MetadataToFhirConvertor` takes in the following arguments:
- `workspace_name`: Name of the Fabric Workspace
- `solution_name`: Name of the DMH OneLake workload solution
- `admin_lakehouse_name`: The lakehouse name where the administration configurations are located
- `one_lake_endpoint` (str): The one lake endpoint. Default is `onelake.dfs.fabric.microsoft.com`
