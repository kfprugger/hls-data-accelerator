##### Calling the FHIRExportService

Using the bulk $export API  provided by Azure Health Data Services, we export FHIR Data on a recurring basis to an Azure Storage Container. The FHIRExportService class facilitates the status monitoring process of these exports.

Here's how the monitoring process works:

**Export Function Key Extraction:** Before initiating an export, the service extracts the ExportFunctionKey secret from a user-specified key vault.

**Triggering the Azure Function:** Using the extracted ExportFunctionKey secret, the FHIRExportService triggers the designated Azure function.

**Continuous Polling:** Once the bulk export has been requested successfully, the URL in the Content-Location header can be used to periodically poll for the status of the export.

**Completion Confirmation:** The polling process continues until the server returns a 200 HTTP status code. This is returned by the server when the export operation has completed. It could have partially succeeded.
