# Overview

This notebook leverages the `TransformToFHIRNDJSON` function from the Healthcare Data Solutions library to convert claims data from the respective delta tables into various FHIR resources—such as Claims and ExplanationOfBenefit—in the Bronze Fhir4.3 format. The output is saved as ndjson files. 

## TransformToFHIRNDJSON
The `TransformToFHIRNDJSON` takes in the following arguments:
- `workspace_name`: Name of the Fabric Workspace
- `solution_name`:  DMH OneLake workload solution name
- `inline_params` : Json dictionary of parameters(configuration values)
- `admin_lakehouse_name`: The lakehouse name of where the administration configurations are located
- `one_lake_endpoint` (str): The one lake endpoint. Default is `onelake.dfs.fabric.microsoft.com`
