# Overview
In this notebook, we will leverage the Healthcare Data Solutions library to move/extract (extract ZIP files and move) files for multiple HDS modalities, which contain various namespaces and file extensions. This framework will move/extract (extract ZIP files and move) files based on the 'file_orchestration_config' configuration, which contains mapping details such as modality, modality_format, extension etc.

## HDS Raw File Orchestration Service
The `RawFileOrchestrationService` takes in the following arguments:
- `workspace_name`: Name of the Fabric Workspace
- `lakehouse_name`: The lakehouse name where the target table and dcm files are located
- `kwargs` (dict): An optional dictionary that customer can use to configure the imaging raw data movement service service
    - `one_lake_endpoint` (str): The one lake endpoint. Default is `onelake.dfs.fabric.microsoft.com`
    - `root_file_ingestion_path` (str): the root of the source path where input (.zip or supported file extension) files will be located. Default will be 'abfss://{workspace_name}@{one_lake_endpoint}/{lakehouse_name}.Lakehouse/Files/Ingest'
    - `root_target_file_path` (str): the root of the target path where input (.zip or supported file extension) files will be located. Default will be 'abfss://{workspace_name}@{one_lake_endpoint}/{lakehouse_name}.Lakehouse/Files/Process'
    - `max_thread_to_extract_archive_file` (int) : Input for max thread size to extract files with in a archive file, otherwise default it from constants
    - `max_thread_to_move_file` (int) : Input for max thread size to move files, otherwise default it from constants
    - `max_archive_file_size_in_mb` (int): Input for max archive file size in MB, otherwise default it from constants
    - `modality` (list): The hds data modality to be processed, Default will be all the modalities in the file orchestration config. example:['Clinical']
    - `modality_format` (list): The format of the modality data to be processed, Default will be all the formats in the file orchestration config. example:['FHIR-NDJSON']
